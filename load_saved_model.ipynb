```python
import joblib
import pandas as pd

saved_package = joblib.load("predictive_maintenance_model.joblib")

model = saved_package["model"]
scaler = saved_package["scaler"]
feature_names = saved_package["feature_names"]
```

```python
new_machine = pd.DataFrame([{
    "Type": 1,  # L = 0, M = 1, H = 2
    "Air temperature [K]": 300.0,
    "Process temperature [K]": 310.0,
    "Rotational speed [rpm]": 1500,
    "Torque [Nm]": 45.0,
    "Tool wear [min]": 100
}])
```

```python
new_machine["temperature_difference"] = (
    new_machine["Process temperature [K]"]
    - new_machine["Air temperature [K]"]
)

new_machine["power_proxy"] = (
    new_machine["Torque [Nm]"]
    * new_machine["Rotational speed [rpm]"]
)

new_machine["wear_torque"] = (
    new_machine["Tool wear [min]"]
    * new_machine["Torque [Nm]"]
)
```

```python
new_machine = new_machine[feature_names]

new_machine_scaled = scaler.transform(new_machine)

failure_probability = model.predict_proba(new_machine_scaled)[0][1]

print("Failure probability:", round(failure_probability * 100, 2), "%")
```

```python
if failure_probability >= 0.70:
    print("High risk: perform urgent preventive maintenance.")
elif failure_probability >= 0.35:
    print("Medium risk: schedule inspection within 7 days.")
else:
    print("Low risk: continue normal monitoring.")
```
